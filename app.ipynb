```python
import nltk
from nltk.stem.porter import PorterStemmer

nltk.download('punkt')  # for tokenizing

stemmer = PorterStemmer()



```

```python
def stem_text(text):
    return " ".join([stemmer.stem(word) for word in nltk.word_tokenize(text.lower())])

df['tags'] = df['tags'].apply(stem_text)
```

```python
from nltk.corpus import stopwords

import nltk
# nltk.download('stopwords')  # only first time

stop_words = stopwords.words("english")  # ✅ correct
```

```python
import nltk
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer

nltk.download('stopwords')
stop_words = stopwords.words('english')

cv = CountVectorizer(stop_words=stop_words)
X = cv.fit_transform(df["tags"])

```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(X,dense_output=False)

print(similarity_matrix.toarray())
```

```python
df["book_title"] = df["book_title"].apply(lambda x: " ".join(x))
```

```python
movie_index = {book: i for i, book in enumerate(df['book_title'])}
```

```python
def recommend(book_title, top_n=5):
    # Check if movie exists
    idx = movie_index.get(book_title)
    if idx is None:
        return f"Movie '{book_title}' not found."

    # Get similarity scores for this movie
    sim_scores = list(enumerate(similarity_matrix[idx].toarray()[0]))  # convert sparse row to array

    # Sort by similarity (highest first) and skip itself
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)[1:top_n+1]

    # Get movie indices
    book_indices = [i[0] for i in sim_scores]

    # Return movie names
    return df[['book_title','author']].iloc[book_indices].values.tolist()

```

```python
recommend("The Midnight Library",10)
```
