# Text Preprocessing

````python
import string

english_letters = set(string.ascii_letters)

def has_non_english_letter(text):
    if pd.isna(text):
        return False

    for char in str(text):
        if char.isalpha() and char not in english_letters:
            return True

    return False
````

# Text Vectorization

````python
def create_book_text(row):
    author = str(row["author"])
    genre = str(row["genres"])
    title = str(row["book_title"])
    details = str(row["book_details"])

    return (
        (author + " ") * 6 +
        (genre + " ") * 3 +
        title + " " +
        details
    )

df["combined_text"] = df.apply(create_book_text, axis=1)
````

## How TF-IDF Works

**TF-IDF (Term Frequency–Inverse Document Frequency)** converts text into numerical vectors based on how important each word is within a document and across all documents.

### Formula

$$
TF\text{-}IDF = TF \times IDF
$$

* **TF (Term Frequency):** Measures how often a word appears in a document.
* **IDF (Inverse Document Frequency):** Gives lower importance to common words and higher importance to rare words.

In a book recommendation system, TF-IDF converts each book's **author, genre, title, and details** into a numerical vector. These vectors can then be compared using **cosine similarity** to find similar books.

```text
Book Information
      ↓
     TF-IDF
      ↓
Numerical Vectors
      ↓
Cosine Similarity
      ↓
Similar Books
```

By repeating important fields such as **author** and **genre**, their terms can have greater influence on the resulting TF-IDF representation.


````python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2),
    max_features=50000
)

tfidf_matrix = tfidf.fit_transform(df["combined_text"])
````

### How Cosine Similarity Works

Cosine similarity measures how similar two vectors are by calculating the **angle between them**.

In a book recommendation system, each book is converted into an **embedding vector**. Cosine similarity compares these vectors to find books with similar semantic meaning.

### Formula

$$
\text{Cosine Similarity}(A,B)
=
\frac{A \cdot B}{\|A\|\|B\|}
$$

### Scores

* **1** → Very similar
* **0** → Little or no similarity
* **-1** → Opposite directions

The system calculates similarity between the selected book and other books, then recommends those with the **highest similarity scores**.


````python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(tfidf_matrix)

print(similarity_matrix)
````

````python
def keep_top_k(sim_matrix, k=10, exclude_self=True):
    """
    Keep only the top-k similarities for each row.
    
    Parameters
    ----------
    sim_matrix : np.ndarray
        Square similarity matrix (books × books)
    k : int
        Number of most similar books to keep
    exclude_self : bool
        Whether to ignore the diagonal (self-similarity = 1.0)
    
    Returns
    -------
    np.ndarray
        Sparse-like dense matrix with only top-k values kept
    """
    sim = sim_matrix.copy()          # work on a copy
    n = sim.shape[0]

    if exclude_self:
        np.fill_diagonal(sim, -np.inf)   # temporarily kill self-similarity

    # Get the indices of the top-k values in each row
    # argpartition is much faster than full argsort for large matrices
    top_k_idx = np.argpartition(sim, -k, axis=1)[:, -k:]

    # Create a mask of zeros
    mask = np.zeros_like(sim, dtype=bool)

    # Mark the top-k positions as True
    rows = np.arange(n)[:, None]
    mask[rows, top_k_idx] = True

    # Keep original values only where mask is True
    result = np.where(mask, sim_matrix, 0.0)

    # Restore the diagonal if you want self-similarity kept
    if exclude_self:
        np.fill_diagonal(result, 1.0)   # or 0.0 if you prefer

    return result
````

````python
similarity_matrix = keep_top_k(similarity_matrix, k=10, exclude_self=True)
````
