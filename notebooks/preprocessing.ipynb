```python
import numpy as np
import pandas as pd
import re

df = pd.read_csv(r"dataset/Book_Details.csv")
df
```

```python
import string

english_letters = set(string.ascii_letters)

def has_non_english_letter(text):
    if pd.isna(text):
        return False

    for char in str(text):
        if char.isalpha() and char not in english_letters:
            return True

    return False
```

```python
mask = df["book_title"].apply(has_non_english_letter)
```

```python
df[mask].head(20)
```

```python
mask = df["book_title"].apply(has_non_english_letter)

df = df[~mask].copy()
```

```python
df
```

```python
def clean_genres(x):
    if pd.isna(x):
        return []
    
    x = str(x).strip("[]")
    
    return [
        genre.strip().strip("'\"")
        for genre in x.split(",")
        if genre.strip()
    ]

df['genres'] = df['genres'].apply(clean_genres)
df = df.explode('genres')
```

```python
df
```

```python
# Find the 20 most frequent genres
top_20_genres = df['genres'].value_counts().head(20).index

# Replace all other genres with "Other"
df['genres'] = df['genres'].where(
    df['genres'].isin(top_20_genres),
    'Other'
)
```

```python
df = df.sort_values(
    ['genres', 'average_rating'],
    ascending=[True, False]
)
```

```python
df
```

```python
df['genre_rank'] = df.groupby('genres').cumcount() + 1
```

```python
df = df.sort_values(
    ['genre_rank', 'genres']
)
```

```python
df['genres'].unique()
```

```python
homepage = df.sort_values(
    ["genre_rank", "genres"]
)
```

```python
df["genres"] = df.groupby(
    ["book_title", "author"]
)["genres"].transform(
    lambda x: ", ".join(x.dropna().astype(str).unique())
)

df = df.drop_duplicates(
    subset=["book_title", "author"]
).reset_index(drop=True)
```

```python
df.to_csv("dataset/books_cleaned.csv", index=False)
```
